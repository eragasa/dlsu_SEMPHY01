## Problem Statement
The variation of silicon and GaAs bandgaps with temperature can be expressed as
$$
E_g(T) = E_g(0) - \frac{\alpha T^2}{T + \beta},
$$

where:

- For **silicon**:  
  $E_g(0) = 1.17\ \text{eV}$,  
  $\alpha = 4.73 \times 10^{-4}\ \text{eV/K}$,  
  $\beta = 636\ \text{K}$.

- For **GaAs**:  
  $E_g(0) = 1.519\ \text{eV}$,  
  $\alpha = 5.405 \times 10^{-4}\ \text{eV/K}$,  
  $\beta = 204\ \text{K}$.

The bandgap ($E_g$) variation with respect to to temperature $T$
$$
E_g(T) = E_g(0) - \frac{\alpha T^2}{T + \beta},
$$
with $\alpha$ and $\beta$ are phenomenalogical fitting parameters


```python

# input: T: array
# parameters alpha, beta are parameters

def get_bandgap_varshni1970(
	T: float, 
	Eg_0: float,
	alpha: float, 
	beta: float) -> float:
  """
  T (float): temperature, in K
  Eg_0: bandgap at T=0
  alpha: fitting paramaeter
  beta: fitting parameter
  
  Reference:
    Varshni., Phys. Rev. B. 2.10 (1970): 3952.
  """
  
  E_g = Eg_0 - ((alpha*T**2)/(T+beta))
  return E_g
```
