```python
import numpy as np

# --- Physical/num params ---
hbar = 1.0
m    = 1.0
N    = 2**12
L    = 200.0
dx   = L / N
x    = (np.arange(N) - N//2) * dx

dk   = 2*np.pi / L
k    = (np.arange(N) - N//2) * dk
k    = np.fft.fftshift(k)

# --- Initial Gaussian packet ---
x0   = -40.0
k0   =  1.0
sigma0 = 5.0

def create_gaussian_packet(x, x0, k0, sigma0):
  norm = (1/(2*np.pi*sigma0**2))**0.25
  psi0 = norm * np.exp(- (x - x0)**2/(4*sigma0**2) + 1j*k0*(x - x0))
  return psi0



# --- Time evolution via spectral method ---
def evolve_free(psi_x, t):
    # to k-space
    psi_k = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(psi_x))) * dx/np.sqrt(2*np.pi)
    phase = np.exp(-1j * (hbar*k**2/(2*m)) * t)
    psi_k *= phase
    # back to x-space
    psi_t = np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(psi_k))) * np.sqrt(2*np.pi)/dx
    # renormalize (guard against roundoff)
    psi_t /= np.sqrt(np.sum(np.abs(psi_t)**2)*dx)
    return psi_t

# example usage
t = 60.0
psi_t = evolve_free(psi0, t)

# diagnostics
prob = np.sum(np.abs(psi_t)**2)*dx
x_mean = np.sum(x*np.abs(psi_t)**2)*dx
p_mean = hbar*np.sum(k*np.abs(np.fft.fftshift(np.fft.fft(np.fft.ifftshift(psi_t))) * dx/np.sqrt(2*np.pi))**2)*dk
```
