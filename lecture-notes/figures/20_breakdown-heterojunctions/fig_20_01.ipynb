# Breakdown and heterojunctions

Quadrature and root finding convert a local ionization law into breakdown voltage; the electron-affinity rule supplies an ideal band lineup.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path.cwd()
while not (ROOT / "figures" / "src").exists():
    if ROOT.parent == ROOT:
        raise RuntimeError("Could not locate lecture-notes/figures/src")
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "figures" / "src"))
from scipy.integrate import trapezoid
from scipy.constants import elementary_charge as q, epsilon_0

eps=11.7*epsilon_0; ND=1e21; Vbi=0.8; VR=100.0; Aion=7e7; Bion=1.23e8
W=np.sqrt(2*eps*(Vbi+VR)/(q*ND)); Emax=q*ND*W/eps
x=np.linspace(0,W,2001); E=Emax*(1-x/W)
alpha=np.zeros_like(E); alpha[:-1]=Aion*np.exp(-Bion/E[:-1])
integral=trapezoid(alpha,x)
print(f"ionization integral at {VR:.0f} V = {integral:.6f}")
```

```python
from semphy.junctions import avalanche_breakdown_voltage, avalanche_integral

Vbr=avalanche_breakdown_voltage(ND,Vbi,eps,Aion,Bion)
criterion=avalanche_integral(Vbr,ND,Vbi,eps,Aion,Bion)
print(f"breakdown voltage = {Vbr:.4f} V; integral = {criterion:.12f}")
assert abs(criterion-1)<1e-9
```

```python
from semphy.junctions import anderson_band_offsets

ec,ev=anderson_band_offsets(4.07,1.42,3.50,2.16)
identity=(ec-ev)-(2.16-1.42)
print(f"ΔEc = {ec:.3f} eV; ΔEv = {ev:.3f} eV; identity residual = {identity:.3e} eV")
assert abs(identity)<1e-14
```
