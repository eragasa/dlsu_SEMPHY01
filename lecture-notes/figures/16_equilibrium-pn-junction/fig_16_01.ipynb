# Equilibrium p–n junction

Direct depletion electrostatics precedes the reusable nonlinear Poisson–Boltzmann solver.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path.cwd()
while not (ROOT / "figures" / "src").exists():
    if ROOT.parent == ROOT:
        raise RuntimeError("Could not locate lecture-notes/figures/src")
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "figures" / "src"))
from scipy.constants import elementary_charge as q, Boltzmann as k_B, epsilon_0

NA, ND, ni, T = 5e22, 1e22, 1e16, 300.0
eps = 11.7 * epsilon_0
VT = k_B*T/q
Vbi = VT*np.log(NA*ND/ni**2)
W = np.sqrt(2*eps*Vbi/q*(1/NA+1/ND))
xp, xn = ND/(NA+ND)*W, NA/(NA+ND)*W
print(f"V_bi = {Vbi:.6f} V; W = {W*1e6:.4f} µm")
print(f"charge-neutrality relative error = {abs(NA*xp-ND*xn)/(ND*xn):.3e}")
```

```python
from semphy.junctions import abrupt_depletion_profile, solve_poisson_boltzmann_junction

dep = abrupt_depletion_profile(NA, ND, ni, T, eps)
pb = solve_poisson_boltzmann_junction(NA, ND, ni, T, eps)
pb_drop = pb["potential_v"][-1] - pb["potential_v"][0]
print(f"Poisson–Boltzmann drop = {pb_drop:.6f} V")
print(f"normalized nonlinear residual = {pb['maximum_residual']:.3e}")
assert abs(pb_drop - Vbi) / Vbi < 1e-8
assert pb["maximum_residual"] < 1e-8
```

```python
# The electrostatic potential must rise monotonically from p bulk to n bulk.
assert np.min(np.diff(pb["potential_v"])) > -1e-10
print(f"peak depletion field = {dep['maximum_field_v_m']:.3e} V/m")
```
