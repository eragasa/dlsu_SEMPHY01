# Figure 6.5 — Intrinsic carrier concentration versus temperature

In the nondegenerate parabolic-band model, $n_i=\sqrt{N_cN_v}e^{-E_g/(2k_BT)}$ and $\mu_i$ follows from intrinsic charge neutrality.

```python
import numpy as np

k_b = 8.617333262e-5
temperature = np.array([200.0, 300.0, 400.0])
energy_gap = 1.12
n_c_300, n_v_300 = 2.8e19, 1.04e19
n_c = n_c_300 * (temperature / 300.0)**1.5
n_v = n_v_300 * (temperature / 300.0)**1.5
n_i = np.sqrt(n_c * n_v) * np.exp(-energy_gap / (2.0 * k_b * temperature))
mu_i = energy_gap / 2.0 + 0.5 * k_b * temperature * np.log(n_v / n_c)
print('T (K):', temperature)
print('n_i (cm^-3):', n_i)
print('mu_i-E_v (eV):', mu_i)
```
