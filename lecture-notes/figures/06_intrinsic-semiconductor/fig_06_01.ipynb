# Figure 6.1 — From atomic levels to crystal bands

Coupling among repeated atomic orbitals produces an increasing number of closely spaced one-electron levels. In the macroscopic limit these families form bands.

```python
import numpy as np

for atom_count in (1, 8, 80):
    lower = np.linspace(0.45, 0.95, atom_count)
    upper = np.linspace(2.35, 2.85, atom_count)
    print(f'{atom_count:2d} atoms -> {lower.size + upper.size:3d} illustrated levels')
```
