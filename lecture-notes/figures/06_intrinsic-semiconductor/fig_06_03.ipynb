# Figure 6.3 — Available states, occupation, and carrier population

The electron spectral population is $g_c(E)f(E)$ and the hole spectral population is $g_v(E)[1-f(E)]$. Integrating them gives $n$ and $p$.

```python
import numpy as np

k_b_t = 0.02585
energy = np.linspace(-0.5, 1.6, 50001)
mu = 0.55
fermi = 1.0 / (1.0 + np.exp(np.clip((energy - mu) / k_b_t, -700, 700)))
g_c = np.sqrt(np.maximum(energy - 1.1, 0.0))
g_v = np.sqrt(np.maximum(0.0 - energy, 0.0))
n = np.trapezoid(g_c * fermi, x=energy)
p = np.trapezoid(g_v * (1.0 - fermi), x=energy)
print('illustrative electron integral:', n)
print('illustrative hole integral:', p)
```
