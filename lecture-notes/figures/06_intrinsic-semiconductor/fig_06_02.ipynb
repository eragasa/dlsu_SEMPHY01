# Figure 6.2 — Reciprocal-space state counting

Periodic boundary conditions give $\Delta k=2\pi/L$. Each allowed wave vector represents one spatial state per band and two states when spin degeneracy is included.

```python
import numpy as np

cell_count, lattice_constant = 24, 0.543e-9
length = cell_count * lattice_constant
delta_k = 2.0 * np.pi / length
k = 2.0 * np.pi * np.fft.fftfreq(cell_count, d=lattice_constant)
print('Delta k =', delta_k, 'm^-1')
print('wave vectors in first-zone grid:', k.size)
print('spin-resolved states:', 2 * k.size)
```
