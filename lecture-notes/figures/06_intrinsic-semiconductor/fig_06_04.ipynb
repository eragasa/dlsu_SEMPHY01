# Figure 6.4 — Fermi–Dirac and Boltzmann-tail occupations

The Boltzmann form $e^{-(E-\mu)/(k_BT)}$ is the high-energy tail of the exact Fermi–Dirac occupation.

```python
import numpy as np

x = np.array([1.0, 2.0, 3.0, 5.0, 8.0])
fermi = 1.0 / (1.0 + np.exp(x))
boltzmann = np.exp(-x)
relative_error = (boltzmann - fermi) / fermi
for value, error in zip(x, relative_error, strict=True):
    print(f'(E-mu)/kT={value:3.0f}: relative error={error:.6f}')
```
