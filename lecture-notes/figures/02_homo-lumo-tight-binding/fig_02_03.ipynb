# Figure 2.3 — Gap formation in a dimerized chain

For alternating couplings, $E_\pm(k)=\epsilon\pm|t_1+t_2e^{-ika}|$. The zone-boundary gap is $2|t_1-t_2|$.

```python
import numpy as np

k_a = np.linspace(-np.pi, np.pi, 20001)
for t_1, t_2 in ((1.0, 1.0), (1.3, 0.7)):
    magnitude = np.abs(t_1 + t_2 * np.exp(-1j * k_a))
    numerical_gap = 2.0 * magnitude.min()
    print((t_1, t_2), 'numerical gap =', numerical_gap, 'analytic gap =', 2.0 * abs(t_1 - t_2))
```
