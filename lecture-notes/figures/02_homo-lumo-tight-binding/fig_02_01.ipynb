# Figure 2.1 — Two-site bonding and antibonding levels

Diagonalizing the two-site Hamiltonian produces the bonding and antibonding energies $\epsilon\mp t$.

```python
import numpy as np

epsilon, t = 0.0, 1.0
hamiltonian = np.array([[epsilon, -t], [-t, epsilon]])
energies, orbitals = np.linalg.eigh(hamiltonian)
print('energies:', energies)
print('normalized orbitals by column:\n', np.round(orbitals, 6))
print('splitting:', energies[1] - energies[0])
```
