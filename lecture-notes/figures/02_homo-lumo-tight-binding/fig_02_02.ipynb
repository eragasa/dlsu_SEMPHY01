# Figure 2.2 — Finite molecular levels approaching a band

An open chain of $N$ equivalent orbitals has $E_m=\epsilon-2t\cos[m\pi/(N+1)]$. Increasing $N$ fills the interval $[\epsilon-2t,\epsilon+2t]$.

```python
import numpy as np

for site_count in (2, 5, 12, 50):
    m = np.arange(1, site_count + 1)
    energies = -2.0 * np.cos(m * np.pi / (site_count + 1))
    print(f'N={site_count:2d}: levels={len(energies):2d}, range=({energies.min():.6f}, {energies.max():.6f})')
```
