# Figure 1.3 — Bloch factorization

For $\psi_k(x)=e^{ikx}u_k(x)$ and $u_k(x+a)=u_k(x)$, translation by one cell gives $\psi_k(x+a)=e^{ika}\psi_k(x)$.

```python
import numpy as np

a = 1.0
k = 0.55 * np.pi / a
x = np.linspace(0.0, 5.0 * a, 10001)
u = lambda position: 1.0 + 0.38 * np.cos(2.0 * np.pi * position / a)
psi = lambda position: np.exp(1j * k * position) * u(position)
translation_error = np.max(np.abs(psi(x + a) - np.exp(1j * k * a) * psi(x)))
periodicity_error = np.max(np.abs(u(x + a) - u(x)))
print('periodicity error:', periodicity_error)
print('Bloch translation error:', translation_error)
```
