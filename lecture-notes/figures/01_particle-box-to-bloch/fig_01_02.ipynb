# Figure 1.2 — Hard-wall and periodic boundary conditions

Hard walls select $k_n=n\pi/L$ with zero endpoint amplitudes. Periodicity selects $k_m=2\pi m/L$ with equal endpoint values.

```python
import numpy as np

L = 1.0
x = np.linspace(0.0, L, 10001)
hard_wall = np.sqrt(2.0 / L) * np.sin(3.0 * np.pi * x / L)
periodic = np.exp(1j * 4.0 * np.pi * x / L)
print('hard-wall endpoints:', hard_wall[[0, -1]])
print('periodic endpoint mismatch:', abs(periodic[-1] - periodic[0]))
```
