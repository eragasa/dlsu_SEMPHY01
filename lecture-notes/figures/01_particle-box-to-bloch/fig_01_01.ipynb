# Figure 1.1 — Particle-in-a-box states and spectrum

The hard-wall eigenfunctions are $\phi_n(x)=\sqrt{2/L}\sin(n\pi x/L)$ and the energies obey $E_n/E_1=n^2$. The first calculation evaluates these relations directly.

```python
import numpy as np

x = np.linspace(0.0, 1.0, 20001)
quantum_numbers = np.arange(1, 5)
wavefunctions = np.sqrt(2.0) * np.sin(np.pi * quantum_numbers[:, None] * x)
norms = np.trapezoid(wavefunctions**2, x=x, axis=1)
energy_ratios = quantum_numbers**2
print('normalizations:', np.round(norms, 12))
print('E_n / E_1:', energy_ratios)
```
