# Figure 3.1 — From a Bloch band to its density of states

For $E(k)=\epsilon-2t\cos(ka)$, the one-dimensional DOS follows from the inverse slope $|dE/dk|^{-1}$ and integrates to two spin-resolved states per cell.

```python
import numpy as np

a, epsilon, t = 1.0, 0.0, 1.0
k = np.linspace(-np.pi / a, np.pi / a, 200001, endpoint=False)
energy = epsilon - 2.0 * t * np.cos(k * a)
bins = np.linspace(-2.05, 2.05, 601)
counts, edges = np.histogram(energy, bins=bins)
widths = np.diff(edges)
dos = 2.0 * counts / (len(k) * widths)
print('integrated states per cell:', np.sum(dos * widths))
print('sampled band edges:', energy.min(), energy.max())
```
