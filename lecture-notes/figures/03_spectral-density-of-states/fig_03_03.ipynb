# Figure 3.3 — Reciprocal-space and energy-resolution convergence

A sampled-band DOS depends on both the reciprocal-space grid and the energy broadening. State-count normalization is the first convergence invariant.

```python
import numpy as np

for sample_count in (200, 2000, 20000):
    k = np.linspace(-np.pi, np.pi, sample_count, endpoint=False)
    energy = -2.0 * np.cos(k)
    counts, edges = np.histogram(energy, bins=240)
    widths = np.diff(edges)
    dos = 2.0 * counts / (sample_count * widths)
    print(sample_count, 'samples -> integrated states =', np.sum(dos * widths))
```
