# Figure 3.2 — Dimensionality of a parabolic-band DOS

Near a parabolic band edge, the DOS scales as $(E-E_c)^{d/2-1}$. The exponents are $-1/2$, $0$, and $+1/2$ in one, two, and three dimensions.

```python
import numpy as np

energy = np.array([0.01, 0.04, 0.16])
for dimension in (1, 2, 3):
    exponent = dimension / 2.0 - 1.0
    scaled_dos = energy**exponent
    print(f'd={dimension}, exponent={exponent:+.1f}, relative DOS={scaled_dos}')
```
