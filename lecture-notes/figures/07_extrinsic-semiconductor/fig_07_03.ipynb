# Figure 7.3 — Compensation and majority carriers

Under complete ionization, charge neutrality and mass action give $n-p=N_D-N_A$ and $np=n_i^2$.

```python
import numpy as np

n_i = 1.0e10
net_doping = np.array([-1.0e17, -1.0e14, 0.0, 1.0e14, 1.0e17])
n = 0.5 * (net_doping + np.sqrt(net_doping**2 + 4.0 * n_i**2))
p = n_i**2 / n
print('n-p closure error:', np.max(np.abs((n - p) - net_doping)))
print('np/n_i^2:', n * p / n_i**2)
```
