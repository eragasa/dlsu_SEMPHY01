# Figure 7.1 — Dopant levels and charge states

A donor becomes positively charged after donating its bound electron. An acceptor becomes negatively charged after accepting a valence electron.

```python
import numpy as np

energy_gap = 1.12
donor_binding = 0.045
acceptor_binding = 0.057
donor_level = energy_gap - donor_binding
acceptor_level = acceptor_binding
print('donor level above E_v:', donor_level, 'eV')
print('acceptor level above E_v:', acceptor_level, 'eV')
print('level ordering:', np.array([0.0, acceptor_level, donor_level, energy_gap]))
```
