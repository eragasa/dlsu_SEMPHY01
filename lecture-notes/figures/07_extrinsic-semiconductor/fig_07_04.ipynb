# Figure 7.4 — Freeze-out, extrinsic, and intrinsic regimes

The temperature evolution follows from the same neutrality equation as dopant ionization and intrinsic excitation change their relative importance.

```python
import numpy as np

k_b = 8.617333262e-5
temperature = np.array([50.0, 150.0, 300.0, 600.0])
activation_energy = 0.045
freeze_out_factor = np.exp(-activation_energy / (2.0 * k_b * temperature))
intrinsic_factor = np.exp(-1.12 / (2.0 * k_b * temperature))
print('T (K):', temperature)
print('donor activation scale:', freeze_out_factor)
print('intrinsic excitation scale:', intrinsic_factor)
```
