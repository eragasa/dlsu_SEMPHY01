# Figure 7.2 — Dopant ionization fractions

With explicit degeneracy factors, donor and acceptor ionization fractions are functions of $(\mu-E_D)/(k_BT)$ and $(E_A-\mu)/(k_BT)$.

```python
import numpy as np

x = np.array([-6.0, -2.0, 0.0, 2.0, 6.0])
g_d, g_a = 2.0, 4.0
donor_ionized = 1.0 / (1.0 + g_d * np.exp(x))
acceptor_ionized = 1.0 / (1.0 + g_a * np.exp(-x))
print('scaled chemical-potential offsets:', x)
print('donor ionized fraction:', donor_ionized)
print('acceptor ionized fraction:', acceptor_ionized)
```
